```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
def cikarilan_ve_kalan(once, sonra):
    print("Çıkarılan:", len(once) - len(sonra))
    print("Kalan:", len(sonra))
```

```python
plt.scatter(df["net_sqm"], df["price"] / 1_000_000_000)
plt.xlabel("Net metrekare")
plt.ylabel("Fiyat (Milyar TL)")
plt.show()
```

```python
df_temiz = df[df["price"] <= 1_000_000_000]

cikarilan_ve_kalan(df, df_temiz)
```

```python
df_temiz_v2 = df_temiz[df_temiz["net_sqm"] <= df_temiz["gross_sqm"]]

cikarilan_ve_kalan(df_temiz, df_temiz_v2)
```

```python
plt.scatter(df_temiz_v2["net_sqm"], df_temiz_v2["price"] / 1_000_000_000)
plt.xlabel("Net metrekare")
plt.ylabel("Fiyat (Milyar TL)")
plt.show()
```

```python
bos_yuzde = df_temiz_v2.isna().mean() * 100
bos_yuzde.sort_values(ascending=False)
```

```python
df_temiz_v2 = df_temiz_v2.copy()

floor = df_temiz_v2["floor"]
total_floors = df_temiz_v2["total_floors"]
floor_category = df_temiz_v2["floor_category"]

df_temiz_v2["is_top_floor"] = (floor == total_floors).where(
    floor.notna() & total_floors.notna()
)
df_temiz_v2["is_ground_floor"] = (floor == 0).where(floor.notna())

df_temiz_v2.loc[floor_category == "Top Floor", "is_top_floor"] = True
df_temiz_v2.loc[floor_category == "Ground Floor", "is_ground_floor"] = True
```

```python
df_temiz_v3 = df_temiz_v2.drop(columns=[
    "complex_name",
    "floor_category",
    "maintenance_fee",
    "building_type",
    "fuel_type",
    "building_condition",
    "last_updated",
    "scraped_at",
    "floor"
])
```

```python
df_temiz_v3.isna().sum()
```

```python
df_temiz_v3["is_top_floor"] = df_temiz_v3["is_top_floor"].astype("boolean").fillna(False)
df_temiz_v3["is_ground_floor"] = df_temiz_v3["is_ground_floor"].astype("boolean").fillna(False)
```
